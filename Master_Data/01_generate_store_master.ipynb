```python
import pandas as pd
from pathlib import Path

# Define source-data path
MASTER_PATH = Path(
    "/Volumes/retailpulse/source/source_files/master"
)

# Create folder if it doesn't exist
MASTER_PATH.mkdir(parents=True, exist_ok=True)

# Generate store data
stores = pd.DataFrame([
    ("ST001", "Hyderabad", "Telangana", "Flagship"),
    ("ST002", "Chennai", "Tamil Nadu", "Standard"),
    ("ST003", "Bengaluru", "Karnataka", "Flagship"),
    ("ST004", "Vijayawada", "Andhra Pradesh", "Standard"),
    ("ST005", "Pune", "Maharashtra", "Standard"),
    ("ST006", "Mumbai", "Maharashtra", "Flagship"),
    ("ST007", "Kochi", "Kerala", "Standard"),
    ("ST008", "Coimbatore", "Tamil Nadu", "Express"),
    ("ST009", "Visakhapatnam", "Andhra Pradesh", "Standard"),
    ("ST010", "Mysuru", "Karnataka", "Express"),
], columns=[
    "store_id", "city", "state", "store_type"
])

display(stores)
```

```python
# Data-quality checks
assert len(stores) == 10, "Expected 10 stores"
assert stores["store_id"].is_unique, "Duplicate store IDs"
assert not stores.isnull().any().any(), "Missing values found"

# Write CSV
file_path = MASTER_PATH / "stores.csv"
stores.to_csv(file_path, index=False)

# Verify output
assert file_path.exists(), "CSV file was not created"

print("Validation passed")
print(f"Rows written: {len(stores)}")
print(f"File path: {file_path}")
```

```python
stores_check = pd.read_csv(file_path)

display(stores_check)

assert len(stores_check) == 10
assert stores_check["store_id"].is_unique

print("CSV read-back validation passed")
```
